# Prediction System

## Objective

This notebook demonstrates how the trained machine learning model can classify a new job posting as either Genuine or Fraudulent.

The prediction pipeline consists of:

1. Loading the trained model
2. Loading the TF-IDF vectorizer
3. Cleaning the input text
4. Converting text into TF-IDF features
5. Predicting the class
6. Displaying the prediction

### Import Libraries

```python
import re
import string
import joblib
import pandas as pd
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

### Download NLTK resources

```python
nltk.download("stopwords")

nltk.download("wordnet")
```

### Load the TF-IDF Vectorizer

```python
tfidf = joblib.load("../models/tfidf_vectorizer.pkl")
```

### Load the Best Model

```python
model = joblib.load("../models/random_forest.pkl")
```

### Create Text Cleaning Function

```python
stop_words = set(stopwords.words("english"))

lemmatizer = WordNetLemmatizer()


def clean_text(text):

    text = text.lower()

    text = re.sub(r"http\S+", "", text)

    text = re.sub(r"\d+", "", text)

    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    words = text.split()

    words = [
        lemmatizer.lemmatize(word)
        for word in words
        if word not in stop_words
    ]

    return " ".join(words)
```

### Create Example Job Posting

```python
job_posting = """
Urgently hiring!

Work from home.

Earn $5000 per week.

No experience required.

Limited seats.

Apply immediately.

Click the link below.
"""
```

### Clean the Text

```python
cleaned_text = clean_text(job_posting)

print(cleaned_text)
```

### Convert to TF-IDF

```python
job_vector = tfidf.transform([cleaned_text])
```

### Predict

```python
prediction = model.predict(job_vector)

print(prediction)
```

### Display User-Friendly Result

```python
def predict_job(job_text):

    cleaned = clean_text(job_text)

    vector = tfidf.transform([cleaned])

    prediction = model.predict(vector)

    if prediction[0] == 1:
        return "Fraudulent Job Posting"

    return "Genuine Job Posting"
```

### Test With the Geniune Job

```python
job_posting = """
Google is hiring a Software Engineer.

Requirements:

Bachelor's degree in Computer Science.

3 years of Python experience.

Health insurance included.

Hybrid work.

Apply through the official careers portal.
"""
```

```python
predict_job("""
Amazon is hiring Data Analysts.
Salary based on experience.
Medical insurance.
Remote work available.
""")
```
