# Text Preprocessing

```python
import pandas as pd 

df = pd.read_csv("../data/cleaned_fake_job_postings.csv")
```

```python
text_columns = [
    "title",
    "company_profile",
    "description",
    "requirements"
]

df[text_columns] = df[text_columns].fillna("")
```

```python
# Create one text column with combining important text columns

df["text"] = (
    df["title"] + " " +
    df["company_profile"] + " " +
    df["description"] + " " +
    df["requirements"]
)
```

### Text Cleaning

```python
# Import Required Libraries

import re
import string
```

```python
# Create a cleaning function

def clean_text(text):
    """
    Clean text by:
    1. Converting to lowercase
    2. Removing HTML tags
    3. Removing URLs
    4. Removing punctuation
    5. Removing numbers
    6. Removing extra spaces
    """

    # Convert to lower case
    text = text.lower()

    # Remove HTML tags
    text = re.sub(r"<.*?>", " ", text)

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", " ", text)

    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # Remove numbers
    text = re.sub(r"\d+", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text 
```

```python
# Test the function

sample = df["text"].iloc[0]

print("Original Text:\n")
print(sample)

print("Cleaned Text:\n")
print(clean_text(sample))
```

### Apply the function

```python
df["clean_text"] = df["text"].apply(clean_text)
```

```python
# Compare before and after

print("Original:\n")
print(df["text"].iloc[0])

print("Cleaned:\n")
print(df["clean_text"].iloc[0])
```
